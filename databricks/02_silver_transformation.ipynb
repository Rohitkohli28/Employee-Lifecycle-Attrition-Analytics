```python
bronze_df = spark.table("bronze_employee")

display(bronze_df)
```

```python
print("Total Records:", bronze_df.count())
```

```python
bronze_df.printSchema()
```

```python
from pyspark.sql.functions import col, to_date

silver_df = bronze_df \
    .withColumn("emp_id", col("emp_id").cast("int")) \
    .withColumn("salary", col("salary").cast("int")) \
    .withColumn("join_date", to_date(col("join_date")))
```

```python
silver_df.printSchema()
```

```python
from pyspark.sql.functions import sum, when

silver_df.select([
    sum(when(col(c).isNull(), 1).otherwise(0)).alias(c)
    for c in silver_df.columns
]).show()
```

```python
silver_df.groupBy("emp_id") \
         .count() \
         .filter(col("count") > 1) \
         .show()
```

```python
silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("silver_employee")
```
